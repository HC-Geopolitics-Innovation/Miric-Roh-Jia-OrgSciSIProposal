# Assemble the Summaries into One Table

| Item | Description |
|---|---|
| Goal | Combine the ten shard files of T1.2.1 into the one table the later sections read, and check that the sample is covered: every hearing summarized once, none twice, none missing. Each shard checks only its own slice, so this is where a shard that died at the wall clock would show as a hole. |
| Input | `Interim/sst_Qwen3.5-4B/shards/summaries_<shard>.parquet` from T1.2.1; `hearings_sst.parquet` from D4 for the hearing metadata. |
| Output | `Processed/hearing_summaries.parquet` in the T1.2 scratch folder: 2,352 rows, the hearing metadata, the run measurements, and `summary`. |
| Variables | • `summary`: the hearing summary written by the reduce call<br>• `summary_tokens`, `n_chunks`, `n_calls`, `n_collapse_rounds`, `seconds`: what the hearing cost<br>• `summarized`: the run finished with status `ok` |
| Run | Last run 2026-08-21; the cell outputs shown are from that run. Not rerun: this table is what T2 and T3 read. It now sits in the Processed folder named above. |

## Step 1. Import packages

```python
from pathlib import Path

import pandas as pd
```

## Step 2. Define file locations

```python
PROJECT_FOLDER = Path.cwd().resolve().parents[2]  # <pipeline>/<section>/<subfolder>/ sits three levels below the clone
SCRATCH_FOLDER = Path("/scratch1/acroh/projects/LLM-Text-Analysis-and-Synthesis")  # the one line to change on another machine
NOTEBOOK_SUBFOLDER = Path("T - Three Step Text Analysis Pipeline/T1 - Text Summarization/T1.2 - Process Entire Sample Congressional Hearings")
NOTEBOOK_NAME = "T1.2.2 - Production - Assemble the Summaries"
INTERIM_FOLDER = SCRATCH_FOLDER / NOTEBOOK_SUBFOLDER / "Interim"
PROCESSED_FOLDER = SCRATCH_FOLDER / NOTEBOOK_SUBFOLDER / "Processed"
INPUT_FOLDER = SCRATCH_FOLDER / "D - Data Collection and Processing/D4 - SST Sample Construction" / "Processed"

PROCESSED_FOLDER.mkdir(parents=True, exist_ok=True)
```

## Step 3. Name the run

```python
MODEL = "Qwen/Qwen3.5-4B"
RUN_NAME = f"sst_{MODEL.split('/')[-1]}"
N_SHARDS = 10

RUN_FOLDER = INTERIM_FOLDER / RUN_NAME
TARGET = PROCESSED_FOLDER / "hearing_summaries.parquet"

print(f"run    : {RUN_NAME}")
print(f"shards : {RUN_FOLDER / 'shards'}")
print(f"output : {TARGET}")
```

## Step 4. Read the shards

A missing shard file means a task never started or never wrote. That is reported here, not discovered later as a gap in the row count.

```python
shard_files = sorted((RUN_FOLDER / "shards").glob("summaries_*.parquet"))
print(f"{len(shard_files)} of {N_SHARDS} shard files present")

missing_shards = set(range(N_SHARDS)) - {int(p.stem.split("_")[-1]) for p in shard_files}
if missing_shards:
    print(f"  MISSING SHARDS: {sorted(missing_shards)}")

frames = [pd.read_parquet(p) for p in shard_files]
summaries = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
for p, f in zip(shard_files, frames):
    ok = int((f["status"] == "ok").sum())
    print(f"  {p.name:<26} {len(f):>5,} rows, {ok:>5,} ok, {f['seconds'].sum() / 3600:>5.2f} h")
print(f"\ntotal {len(summaries):,} rows")
```

## Step 5. Join the hearing metadata

Topic extraction groups by committee and by year, so those columns travel with the text instead of being joined again by every later stage.

```python
META = ["packageId", "title", "congress", "chamber", "year", "heldDate_first",
        "committee_ids", "committee_name", "committee_scope",
        "n_members", "n_witnesses", "n_words", "n_turns"]
sst = pd.read_parquet(INPUT_FOLDER / "hearings_sst.parquet", columns=META)

merged = sst.merge(summaries.drop(columns=["n_words", "n_turns"]), on="packageId", how="left", validate="one_to_one")
merged["summarized"] = merged["status"].eq("ok")

print(f"sample     {len(sst):>6,}")
print(f"summarized {int(merged['summarized'].sum()):>6,} ({merged['summarized'].mean():.1%})")
print(f"failed     {int(merged['status'].notna().sum() - merged['summarized'].sum()):>6,}")
print(f"never run  {int(merged['status'].isna().sum()):>6,}")
```

## Step 6. Write the table

```python
merged = merged.sort_values(["year", "packageId"]).reset_index(drop=True)
merged.to_parquet(TARGET, compression="zstd", index=False)

size = TARGET.stat().st_size
print(f"wrote {TARGET}")
print(f"  {len(merged):,} rows, {size / 1e6:.1f} MB zstd, {len(merged.columns)} columns")
good = merged.loc[merged["summarized"]]
summary_words = good["summary"].str.split().str.len().sum()
print(f"  {summary_words:,} summary words from {good['n_words'].sum() / 1e6:.1f}M transcript words ({summary_words / good['n_words'].sum():.3%})")
print(f"  GPU time: {merged['seconds'].sum() / 3600:.1f} h")
```

## Step 7. Check the table

- One row per hearing, no duplicates.
- Every hearing was attempted by exactly one shard.
- All ten shards contributed.
- Failures are listed.
- Every summary is non-empty and within the 512-token cap.
- Coverage by committee and decade, and the collapse rounds used.

```python
back = pd.read_parquet(TARGET)

assert len(back) == len(sst), f"{len(back):,} rows against {len(sst):,} in the sample"
assert not back["packageId"].duplicated().any(), "a hearing appears twice"
print(f"1) one row per hearing, {len(back):,}, no duplicates")

covered = back.loc[back["status"].notna()]
gap = len(back) - len(covered)
assert gap == 0, f"{gap:,} hearings were never attempted: a shard did not finish. Submit the array again; T1.2.1 resumes from its logs."
print("2) every hearing was attempted by exactly one shard")

per_shard = covered.groupby("shard").size()
assert len(per_shard) == N_SHARDS, f"expected {N_SHARDS} shards, found {len(per_shard)}"
print(f"3) all {N_SHARDS} shards contributed ({per_shard.min():,}-{per_shard.max():,} hearings each)")

failed = back.loc[back["status"].notna() & ~back["summarized"]]
print(f"4) failures: {len(failed)}")
for r in failed.head(5).itertuples():
    print(f"     {r.packageId}: {r.status[:90]}")

good = back.loc[back["summarized"]]
assert (good["summary"].str.strip() != "").all(), "an ok row has no summary"
assert (good["summary_tokens"] <= 512).all(), "a summary exceeds the 512-token cap"
print(f"5) summaries: median {good['summary_tokens'].median():.0f} tokens, max {good['summary_tokens'].max()}, all within the cap")


def as_percent(value):
    return f"{value:.0%}" if pd.notna(value) else "-"


print("\n6) coverage by committee and decade:")
decade = back["year"] // 10 * 10
print(back.pivot_table(index=decade, columns="committee_scope", values="summarized", aggfunc="mean").map(as_percent).to_string())

print("\n7) collapse rounds actually used:")
print(good["n_collapse_rounds"].value_counts().sort_index().to_string())
```
