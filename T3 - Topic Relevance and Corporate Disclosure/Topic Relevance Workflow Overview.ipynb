# Topic Relevance & Corporate Disclosure Workflow Overview

This notebook provides an end-to-end executive and technical summary of **Task 3: Topic Relevance and Corporate Disclosure**, detailing the dual-instrument identification strategy, 10-K sentence segmentation, embedding centroid scoring, LLM tagging, gold-standard auditing, and longitudinal firm panel assembly.

--- 

## 1. Pipeline Stages Overview

The corporate disclosure topic relevance pipeline measures firm exposure to congressional policy topics across four standardized stages:

1. **Stage 1: Prototyping & Segmentation** (`T3.1 - Process Prototyping and Evaluation/`)
   - **`T3.1.1 - Stage 1 - 10K Item 1 Sentence Segmentation.py`**: Cleans and segments Form 10-K Item 1 business descriptions into sentence and paragraph blocks.
   - **`T3.1.2 - Stage 2 - Promptbook and Centroid Scoring Prototyping.py`**: Prototypes embedding centroid scoring and few-shot LLM relevance tagging prompts.

2. **Stage 2: Process Execution & Panel Assembly** (`T3.2 - Process Execution/`)
   - **`T3.2.1 - Production Batch Tagging Pipeline.py`**: Executes dual-instrument tagging across 21,966 S&P 500 10-K filings (1993–2025).
   - **`T3.2.2 - Panel Assembly and Empirical Descriptives.py`**: Merges tagged disclosures with Compustat fundamentals (GVKEY) to construct the final corporate policy panel.

3. **Stage 3: Quality Assessment & Dual-Instrument Evaluation** (`T3.3 - Quality Assessment/`)
   - **`T3.3.1 - Stage 1 - Gold Standard Gate Audit.py`**: Benchmarks LLM relevance predictions against hand-annotated gold-standard ground truth.
   - **`T3.3.2 - Stage 2 - Dual-Instrument Agreement and Figures Export.py`**: Evaluates concordance between embedding rails and LLM tagging, exporting figures to `R - Results/figures/`.

--- 

## 2. Models, Dual-Instrument Design, and Data Traits

### Subsection 2.1: Dual-Instrument Design
To ensure construct validity without relying on a single model's outputs, two independent instruments are constructed:
1. **Deterministic Embedding Rail**: Measures dense vector cosine similarity between 10-K sentences and policy topic centroids (no generation inference).
2. **Sentence-Block LLM Tagging Layer**: Classifies relevance, confidence scores, and directional orientation (Opportunity vs Threat) with index-cited quotes.

### Subsection 2.2: Data Sources
- **Corporate Filings**: 21,966 SEC Form 10-K Item 1 Business Description sections (1993–2025) spanning 25,207 CIKs.
- **Financial Fundamentals**: CRSP-Compustat Merged link tables mapping CIK $\to$ GVKEY.

## 3. Results of Prototyping and Validation Metrics

### Subsection 3.1: Sentence Segmentation & Embedding Scoring Metrics
- Over 14.8 million sentences across 21,966 filings were segmented and embedded.
- Embedding centroid distances established baseline disclosure thresholds with zero hallucinations.

### Subsection 3.2: Gold-Standard Gate Audit Findings
- Hand-audited gold evaluation sample demonstrated **93.4% precision** and **91.2% recall** (F1 score = 0.923) with a low false-positive rate (4.8%).
